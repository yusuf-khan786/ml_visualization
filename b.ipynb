```python
churn = pd.read_csv('D:\\ml_project\\Churn_Modelling.csv')
covid = pd.read_csv('D:\\ml_project\\covid_toy.csv')
```

```python
covid_city = covid.groupby("city").agg(
    total_people=("has_covid", "count"),
    covid_positive=("has_covid", "sum"),
    positive_rate=("has_covid", "mean")
).reset_index()

print(covid_city)
```
